# 구구단 출력기 (Gugudan Printer)

이 코드는 입력받은 숫자에 대한 구구단 (곱셈표)을 출력하는 함수입니다.

```python
def print_gugudan(number):
    """
    주어진 숫자에 대한 구구단 (곱셈표)을 출력합니다.

    Args:
        number (int): 구구단을 출력할 숫자 (1에서 9 사이).
    """
    if not (1 <= number <= 9):
        print("구구단은 1부터 9까지의 숫자만 지원합니다.")
        return

    print(f"--- {number}단 ---")
    for i in range(1, 10):
        print(f"{number} x {i} = {number * i}")

# 7단 구구단 출력 예시
print_gugudan(7)
```

다른 숫자의 구구단을 확인하려면 위의 `print_gugudan()` 함수 호출 부분의 숫자를 변경하여 실행해 보세요.
